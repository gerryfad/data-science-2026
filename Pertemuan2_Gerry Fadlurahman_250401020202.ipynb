```python
import pandas as pd
import numpy as np
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
df.head()
```

### Q1. Berapa jumlah baris dan kolom pada dataset Titanic? Kolom apa saja yang tersedia?
Pada bagian ini saya menampilkan ukuran dataset dan nama seluruh kolom.

```python
print("Jumlah baris dan kolom:", df.shape)
print("Jumlah baris:", df.shape[0])
print("Jumlah kolom:", df.shape[1])
print("Daftar kolom:")
print(df.columns.tolist())
```

### Q2. Kolom mana saja yang memiliki missing values? Berapa jumlah dan proporsinya (%) per kolom?
Pada bagian ini saya mengecek data yang kosong pada setiap kolom.

```python
missing_jumlah = df.isnull().sum()
missing_persen = (df.isnull().mean() * 100).round(2)

hasil_missing = pd.DataFrame({
    "jumlah_missing": missing_jumlah,
    "persentase_missing": missing_persen
})

hasil_missing = hasil_missing[hasil_missing["jumlah_missing"] > 0]
print(hasil_missing)
```

### Q3. Berapa persen penumpang yang selamat (Survived = 1)?
Pada bagian ini saya menghitung persentase penumpang yang selamat.

```python
persen_selamat = df["Survived"].mean() * 100
print(f"Persentase penumpang yang selamat: {persen_selamat:.2f}%")
```

## Q4. Filter penumpang wanita dari kelas 1 (Pclass = 1). Berapa jumlahnya dan berapa rata-rata usianya?
Pada bagian ini saya memfilter penumpang wanita dari kelas 1, lalu menghitung jumlah dan rata-rata usianya.

```python
wanita_kelas1 = df[(df["Sex"] == "female") & (df["Pclass"] == 1)]

jumlah_wanita_kelas1 = wanita_kelas1.shape[0]
rata_usia_wanita_kelas1 = wanita_kelas1["Age"].mean()

print("Jumlah penumpang wanita kelas 1:", jumlah_wanita_kelas1)
print(f"Rata-rata usia penumpang wanita kelas 1: {rata_usia_wanita_kelas1:.2f}")
```

### Q5. Gunakan groupby untuk menghitung tingkat keselamatan (mean Survived) per kelas penumpang. Kelas mana yang paling tinggi?
Pada bagian ini saya menggunakan groupby untuk menghitung tingkat keselamatan pada tiap kelas penumpang.

```python
tingkat_selamat_kelas = df.groupby("Pclass")["Survived"].mean().round(3)
print(tingkat_selamat_kelas)

kelas_tertinggi = tingkat_selamat_kelas.idxmax()
nilai_tertinggi = tingkat_selamat_kelas.max()

print(f"Kelas dengan tingkat keselamatan tertinggi: {kelas_tertinggi}")
print(f"Nilai tingkat keselamatan tertinggi: {nilai_tertinggi:.3f}")
```

### Q6. Buat analisis tambahan berdasarkan variabel lain yang dianggap menarik
Analisis tambahan yang saya pilih adalah membandingkan tingkat keselamatan berdasarkan jenis kelamin.

```python
selamat_berdasarkan_gender = df.groupby("Sex")["Survived"].mean().round(3)
print(selamat_berdasarkan_gender)
```

Berdasarkan hasil di atas, tingkat keselamatan penumpang perempuan lebih tinggi dibandingkan laki-laki. Hal ini menunjukkan bahwa jenis kelamin berpengaruh terhadap peluang keselamatan pada dataset Titanic.
